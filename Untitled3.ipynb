```python
def calculate_average(numbers):
    if not numbers:
        return 0.0

    total = 0

    for number in numbers:
        total += number

    return total / len(numbers)
```

```python
import math


def test_calculate_average():
    test_cases = [
        ([78, 85, 92, 66, 88], 81.8),
        ([], 0.0),
        ([100], 100.0),
        ([-2, 2], 0.0),
        ([1, 2], 1.5),
    ]

    for numbers, expected in test_cases:
        actual = calculate_average(numbers)

        assert math.isclose(
            actual,
            expected,
            rel_tol=1e-9,
            abs_tol=1e-9,
        ), (
            f"输入 {numbers} 时结果错误："
            f"期望 {expected}，实际 {actual}"
        )

    print("第一套题：全部测试通过")


test_calculate_average()
```
