```python
def analyze_scores(scores, pass_score=60):
    for score in scores:
        if score < 0 or score > 100:
            raise ValueError("成绩必须在 0 到 100 之间")

    if not scores:
        return {
            "count": 0,
            "average": 0.0,
            "highest": None,
            "lowest": None,
            "pass_rate": 0.0,
        }
    count = len(scores)
    average = round(sum(scores) / count, 2)
    passed_count = sum(score >= pass_score for score in scores)
    pass_rate = round(passed_count / count * 100, 2)

    return {
        "count": count,
        "average": average,
        "highest": max(scores),
        "lowest": min(scores),
        "pass_rate": pass_rate,
    }
```

```python
# 首次运行应产生 ZeroDivisionError，用于演示可见的失败证据
analyze_scores([])
```

```python
assert analyze_scores([100, 80, 60, 40]) == {
    "count": 4,
    "average": 70.0,
    "highest": 100,
    "lowest": 40,
    "pass_rate": 75.0,
}
assert analyze_scores([]) == {
    "count": 0,
    "average": 0.0,
    "highest": None,
    "lowest": None,
    "pass_rate": 0.0,
}

try:
    analyze_scores([-1])
except ValueError:
    pass
else:
    raise AssertionError("out-of-range score must raise ValueError")

print("Demo tests passed")
```
